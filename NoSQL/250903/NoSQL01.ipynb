----
- db.컬렉션명.find() : 특정 db안에 컬렉션 내부에 어떤 데이터가 있는지 확인
- db.createCollection("컬렉션명") : 해당 db안에 특정 컬렉션을 생성,
- db.createCollection("컬렉션명", options) :
  - capped : true => 고정된 크기를 갖고 있는 컬렉션을 만들겠다 의미
  - autoIndexId : true => _id 필드에 대한 인덱스를 자동으로 생성하겠다
  - size : capped의 옵션 true인 경우에만 사용 => 바이트 단위
  - max : 컬렉션에 저장할 수 있는 문서의 최대 개수를 정의
- db.컬렉션명.drop() : 특정 db안에 컬렉션을 삭제하고자 할 때
  (*만약, db안에 존재하고 있는 컬렉션이 아무것없어질 때, db는 자동으로 삭제)
- db.컬렉션명.isCapped() : 해당 컬렉션의 capped 여부 확인가능!!!
- db.컬렉션명.renameCollection("변경하고 싶은 이름") : 컬렉션 이름 변경

#### 5. MongoDB의 주요 데이터 타입
- String : 문자열 데이터 // name : "David"
- Integer : 정수형 데이터 // age : 20
- Boolean : 논리형 데이터 // isStudent : true 
- Dobule : 실수자료형 데이터 (*부동소수점 형태의 값) // rating : 4.5
- Arrays : 배열 데이터 // 파이썬 내 리스트 자료구조와 비슷하다고 보면 됩니다.
  (*hobbies: ["reading", "music"])
- Object : 객체 데이터 // key & value로 구성되어있는 자료형
  (*address: {city : "Seoul", country : "South Korea"}
- Null : Null 데이터 저장할 수 있는 데이터 형태
  (*Null => 결측치)
- ObjectId : 각각의 문서 ID를 저장하는데 사용되는 타입
- Date : 날짜 데이터 및 시간 정보를 저장하는데 사용되는 타입

#### 8. MongoDB Aggregation Framework
- MongoDB > NoSQL, Aggregation => 집합 | 집계
- Aggregation => 프레임워크
- $push <-> 배열이 아닌 요소들을 하나의 배열의 자료구조 형태로 만들어주는기능
- 프레임워크 | 라이브러리 => 만약
- 기존에 학습했던 find()의 상위호환 버전이라고 생각해도 무방!!!
- Linux 운영체제 => pipeline 개념을 벤치마킹 => 함수의 기능이 구현 => 기능
- shard : 샤드 => A (*1번 샤드) -> B (*2번 샤드)
- Aggregation Framework를 실행하기위한 주요 단계
  - \$addFields : 기존 필드에 추가적으로 속성 더하거나 변경하고자 할 때
  - \$match : 조건에 맞는 문서를 필터링
  - \$unwind : 중첩배열의 값을 하나로 합쳐주는 역할
  - \$group : 문서를 그룹화하고 집계 작업을 수행할 수 있도록 준비
  - \$project : 연산처리된 결과를 특정 요소에 포함시키거나 새로운 문서로 변환 작업
  - \$limit : 특정 개수의 문서를 출력하고자 할 때
  - \$sort : 연산처리된 결과를 정렬 (*위치 무관)
  - \$count : 연산결과값을 통해서 받아온 문서 수의 개수를 집계
  - \$lookup : 서로 다른 컬렉션을 묶어서 편집할 때 사용 (*JOIN)

(*연산처리를 하기 위한 집계 작업 명령어 = 예약어)
- \$sum : 특정 필드명 안에 포함되어있는 값의 합계
- \$avg : 특정 필드명 안에 포함되어있는 값의 평균
- \$min : 특정 필드명 안에 포함되어있는 최소값
- \$max : 특정 필드명 안에 포함되어있는 최대값
- \$first: 특정 필드 안에 첫번째 값
- \$last: 특정 필드 안에 마지막번째 값
- \$push : 기존값들을 배열의 형태로 구성할 때 사용
- \$convert : 특정 필드 내 속성을 변환시킬 때 사용
- \$strLenCP : 문자열의 길이값을 찾아올 때 사용
- \$toString : 문자열 형태가 아닌 타입의 값을 문자로 변환
- \$addToSet : 특정 필드 안에 있는 값을 중복하지 않고 찾아오고자 할 때
- \$concat : 각 필드 안에 존재하는 값을 하나의 문자열로 생성하고자 할 때
