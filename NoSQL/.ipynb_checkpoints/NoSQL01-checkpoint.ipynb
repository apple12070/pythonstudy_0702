#### 1. 데이터베이스 = DB
- 데이터베이스 => 데이터를 보관하고 저장할 수 있는 큰 공간
- 데이터베이스 역사 : 오래되었음
- 데이터를 담고 관리할 수 있는 시스템, 체계의 역사 : 오래되었음
- DBMS 역사 : 오래되었음
- 80여개 정도의 DBMS 존재
- 관계형 데이터베이스 관리 시스템 : RDBMS
- Oracle > MySQL
- SQL -> RDBMS 기반하에 저장.보관되어있는 데이터를 처리하기 위한 목적의 언어
- 스마트폰 -> 모바일 -> 데이터 기하급수적 증가
- 빅데이터 : 하루에도 수백.수천만건의 데이터가 쌓여지는 것을 의미
- RDBMS -> DB -> Table -> Schema
- 빅데이터라는 방대한 데이터의 입.출력건이 수시로 실행되고 있는데, 이것을 과거의 문법 및 방법으로 처리.관리한다는 것이 현실.물리적으로 부담 시작
- 시대의 흐름에 따라서 데이터를 처리할 수 있는 프로그래밍언어 다양해짐
- 각각의 언어가 매번 RDBMS와의 상호작용을 하게끔 처리하는 것도 현실적으로 불가능
- 더이상 SQL 문법만 가지고 RDBMS 라는 데이터 관리 시스템만 의존하는 것은 구시대적인 사고방식!!!
- NoSQL => Not Only SQL
- NoSQL 기반의 데이터 처리 => 전통적인 역사를 가지고 있는 기업 x
- SQL => Oracle > MySQL
- NoSQL => MongoDB

#### 2. NoSQL 이해
- NoSQL 은 스키마 & Join등의 문법 및 관리시스템이 존재하지 x
- 현재 작업하는 업무가 95% read, 5% write : RDBMS 100%
- 현재 작업하는 업무에서 50% 이상 write : NoSQL + SQL

#### 3. MongoDB
- document 기반 NoSQL 데이터베이스
- MongoDB에 저장.관리 되는 데이터가 BSON(Binary JSON)기반의 document 형태로 데이터 관리
- SQL // python => API Data => json, xml
- python =>
  
```
document = {
    "id":"01",
    "language":"java",
    "edition":{
        "first":"1st",
        "second":"2nd",
        "third":"3th"
    }
}
```

```
{
    "_id": ObjectId("504860sdklfjghpan5469327"),
    "username": "david",
    "name": {
        "first": "1st",
        "second": "2nd"
    }
}
```
- SQL : Database - Table(*Schema) - data
- NoSQL : Database - Collection - Document
  
#### 4. 프로그래밍 언어를 보다 효율적으로 학습하는 방법
- CRUD 입각해서 학습
  
- CREATE
- READ
- UPDATE
- DELETE

- show dbs : 로컬컴퓨터 내 생성된 db목록 확인
- use dbname : 특정 db로 접속하겠다는 의미 / 만약 해당 DB가 존재하지 않는 경우 생성
- show collections : 해당 db안에 생성되어있는 컬렉션 목록 확인
- db.stats() : db의 상태정보를 확인
- db.dropDatabase() : db를 삭제하고자 할 때
----
- db.컬렉션명.find() : 특정 db안에 컬렉션 내부에 어떤 데이터가 있는지 확인
- db.createCollection("컬렉션명") : 해당 db안에 특정 컬렉션을 생성,
- db.createCollection("컬렉션명", options) :
  - capped : true => 고정된 크기를 갖고 있는 컬렉션을 만들겠다 의미
  - autoIndexId : true => _id 필드에 대한 인덱스를 자동으로 생성하겠다
  - size : capped의 옵션 true인 경우에만 사용 => 바이트 단위
  - max : 컬렉션에 저장할 수 있는 문서의 최대 개수를 정의
- db.컬렉션명.drop() : 특정 db안에 컬렉션을 삭제하고자 할 때
  (*만약, db안에 존재하고 있는 컬렉션이 아무것없어질 때, db는 자동으로 삭제)
- db.컬렉션명.isCapped() : 해당 컬렉션의 capped 여부 확인가능!!!
- db.컬렉션명.renameCollection("변경하고 싶은 이름") : 컬렉션 이름 변경

#### 5. MongoDB의 주요 데이터 타입
- String : 문자열 데이터 // name : "David"
- Integer : 정수형 데이터 // age : 20
- Boolean : 논리형 데이터 // isStudent : true 
- Dobule : 실수자료형 데이터 (*부동소수점 형태의 값) // rating : 4.5
- Arrays : 배열 데이터 // 파이썬 내 리스트 자료구조와 비슷하다고 보면 됩니다.
  (*hobbies: ["reading", "music"])
- Object : 객체 데이터 // key & value로 구성되어있는 자료형
  (*address: {city : "Seoul", country : "South Korea"}
- Null : Null 데이터 저장할 수 있는 데이터 형태
  (*Null => 결측치)
- ObjectId : 각각의 문서 ID를 저장하는데 사용되는 타입
- Date : 날짜 데이터 및 시간 정보를 저장하는데 사용되는 타입

#### 6.비교연산자
- $eq : =
- $gt : >
- $gte : >=
- $lt : <
- $lte : <=
- $nin : 특정 값을 갖고있지 않은경우
- $in : 특정 값을 갖고있는 경우 (*배열의 자료형태 -> 복수의 값을 기준으로 검색)
- $ne : not enough (*단일 값의 형태)

#### 7.논리연산자
- $or : OR조건
- $and : AND조건
- $not : NOT조건
